```python
df=df.drop_duplicates()
```

```python
#converting negative and unrealstic value to zero using cliping
df['Physical_Activity_Hours']=df['Physical_Activity_Hours'].clip(lower=0)
```

```python
#Skewness
num_col = df.select_dtypes(include='number')
num_col.skew()
# to fix skewness values we cn use log transformation
```

```python
# from here feature enfineering will start
# to selct top 10 categories from any coloumn
top_countries=df['Country'].value_counts().index[:10].tolist()
```

```python
def group_countries(country):
  if country in top_countries:
    return country
  else:
    return "Other"
```

```python
df["Grouped_country"]=df['Country'].apply(group_countries)
```

```python
df["Grouped_country"].value_counts()
df.columns
```

```python
from sklearn.model_selection import train_test_split
skewed_col=['Study_Hours']
other_num_col=["Age","Avg_Daily_Usage_Hours",	"Daily_Unlocks"	,	"Physical_Activity_Hours","Sleep_Hours_Per_Night"]
ordinal_col=['Stress_Level']
normal_col=['Gender','Academic_Level','Most_Used_Platform','Purpose_Of_Use','Grouped_country']
feature_cols=skewed_col+other_num_col+ordinal_col+normal_col
X=df[feature_cols]
y=df['Mental_Health_Score']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=42)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer,StandardScaler,OrdinalEncoder,OneHotEncoder
from sklearn.compose import ColumnTransformer
#1. skewed pipeline
skew_pipeline=Pipeline(steps=[
    ("log_transform",FunctionTransformer(np.log1p)),
    ("scale",StandardScaler())
])
#2 numeric feature
plain_number_pipeline=Pipeline(steps=[
    ("scale",StandardScaler())
])
#3 ordinal
ordinal_pipeline=Pipeline(steps=[
    ("encode",OrdinalEncoder(categories=[['Low','Medium','High','Very High']]))
])
#4 Nominal Feature
nominal_pipeline=Pipeline(steps=[
    ("encode",OneHotEncoder(handle_unknown='ignore'))

])
preprocessor=ColumnTransformer(transformers=[
    ("Skewed_Pipeline",skew_pipeline,skewed_col),
    ("Plain_Numeric",plain_number_pipeline,other_num_col),
    ("Ordinal_pipeline",ordinal_pipeline,ordinal_col),
    ("Normal_pipeline",nominal_pipeline,normal_col)
])
preprocessor
```

```python
#pipeline for models
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score , mean_absolute_error,mean_squared_error
lr_pipeline=Pipeline(steps=[
    ("preprocessor",preprocessor),
    ("Regressoin",LinearRegression())
])
lr_pipeline.fit(X_train,y_train)
lr_preds=lr_pipeline.predict(X_test)
lr_preds_train=lr_pipeline.predict(X_train)
lr_r2_testing=r2_score(y_test,lr_preds)
lr_r2_trainig=r2_score(y_train,lr_preds_train)
print(f'testing accuracy {lr_r2_testing} , training_accuracy {lr_r2_trainig}')
```

```python
from sklearn.ensemble import RandomForestRegressor
Random_forest=Pipeline(steps=[
    ("preprocessor",preprocessor),
    ("Regressoin",RandomForestRegressor())
])
Random_forest.fit(X_train,y_train)
rf_preds=Random_forest.predict(X_test)
rf_preds_train=Random_forest.predict(X_train)
rf_r2_testing=r2_score(y_test,rf_preds)
rf_r2_trainig=r2_score(y_train,rf_preds_train)
print(f'testing accuracy {rf_r2_testing} , training_accuracy {rf_r2_trainig}')
```
